# Kayıp Veri (Missing Value)
**Amaç:**

````python
# yardımcı fonksiyon ile metinleri normlize edelim, bğyk kucuk harf donusumleri, bolslukların kaldırılması, turkce karakter kaynaklı farkların düzeltilmesi

def metin_normalize_et(series):
  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("ş", "s" , regex=False)
      .str.replace("ı", "i" , regex=False)
      .str.replace("ğ", "g" , regex=False)
      .str.replace("ü", "u" , regex=False)
      .str.replace("ç", "c" , regex=False)
      .str.replace("ş", "s" , regex=False)
      .str.replace("ö", "o" , regex=False)
  )
````

# Aykırı Değerler (Outliers)


---


**Amaç:**

1. Veri setindeki aykırı değerlerin tespit edilmesi ve incelenmesi
2. Aykırı değerleri temizleyerek güncel veri setinin oluşturulması

**Adımlar:**

1. Veri setini yükle
2. Sayısal sütunları belirle
3. Sayısal sütunların özet istatistiklerini hesapla
4. IQR ile aykırı değer sınırlarını hesapla
5. Aykırı değer içeren kayıtları tespit etmek
6. Aykırı değer sayılarını sütun bazında incelemek
7. Aykırı değer içeren kayıtları veri setinden temizlemek
8. Temizlenmiş veri setini kaydetmek

````python
# load dataset
df = pd.read_csv("e_ticaret_veri_seti_tekrarlayan_kayitlar_duzenlendi.csv")
df.head()
````

````python
# sayisal sutunların belirlenmesi
sayisal_sutunlar = df.select_dtypes(include = ["int64", "float64"]).columns.tolist()
sayisal_sutunlar
````

````python
# sayisal sutunların özel istatistikleri
df[sayisal_sutunlar].describe()
````

````python


"""birim_fiyat
  IQR = Q3(75%) - Q1(25%)
      = 817 - 198 = 619
  Üst sınır = Q3 + k*IQR
  Alt sınır = Q1 - k*IQR
    k = 1 olsun
    Üst sınır = 817 + 619 = 1436
    Alt sınır = 198 - 619 = -421
  Outlier = üst sınır ve alt sınır dışında kalan örnekler
"""
````

````python
# IQR ile aykırı deger sınırlarını hesaplama
aykiri_sinirlar = {}

for sutun in sayisal_sutunlar:

  q1 = df[sutun].quantile(0.25)
  q3 = df[sutun].quantile(0.75)

  iqr = q3 - q1
  alt_sinir = q1 - 1.5 * iqr
  ust_sinir = q3 + 1.5 * iqr

  aykiri_sinirlar[sutun] = {
      "q1": q1,
      "q3": q3,
      "iqr": iqr,
      "alt_sinir": alt_sinir,
      "ust_sinir": ust_sinir
  }
````

````python
aykiri_sinirlar
````

````python
# aykırı değer içeren kaıtlaı tespit et
aykiri_maskesi = pd.Series(False, index = df.index)

for sutun in sayisal_sutunlar:
  alt_sinir = aykiri_sinirlar[sutun]["alt_sinir"]
  ust_sinir = aykiri_sinirlar[sutun]["ust_sinir"]

  sutun_maskesi = (df[sutun] < alt_sinir) | (df[sutun] > ust_sinir)
  aykiri_maskesi = aykiri_maskesi | sutun_maskesi
````

````python
aykiri_kayitlar = df[aykiri_maskesi]
aykiri_kayitlar
````

````python
# aykırı değer sayılarını sutun bazında ıncele
for sutun in sayisal_sutunlar:

  alt_sinir = aykiri_sinirlar[sutun]["alt_sinir"]
  ust_sinir = aykiri_sinirlar[sutun]["ust_sinir"]

  aykiri_sayi = ((df[sutun] < alt_sinir) | (df[sutun] > ust_sinir)).sum()
  print(f"{sutun}: {aykiri_sayi}")
````

````python
df_temiz = df[~aykiri_maskesi].copy()
df_temiz.shape
````

````python
#temizlenmiş veri setini kaydet
df_temiz.to_csv("e_ticaret_veri_seti_aykiri_degerler_duzenlendi.csv")
````

**Amaç:**

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
df = pd.read_csv("e_ticaret_veri_seti_aykiri_degerler_duzenlendi.csv")
print("ok")
````

````python
# her sutun icin oran analizi yapalım
for sutun in kategorik_sutunlar:
  print((df[sutun].value_counts(normalize=True) * 100).round(2))
  print()
````

````python
# en az görülen kategorileri inceleyelim
for sutun in kategorik_sutunlar:
  print()
  print(df[sutun].value_counts().tail(3))
````

````python
#belirli sütunlar üzerinde detaylı kategorik analiz
detayli_analiz_sutunlari = ["sehir", "kategori","odeme_turu","musteri_tipi"]

for sutun in detayli_analiz_sutunlari:
  if sutun in df.columns:
    print("Frekanslar: ")
    print(df[sutun].value_counts())

    print("Oranlar: ")
    print((df[sutun].value_counts(normalize=True) * 100).round(2))
````

````python
# belirli sutunlar icin countplot çizdirme
for sutun in detayli_analiz_sutunlari:
  plt.figure(figsize=(6,3))
  sns.countplot(data=df, x = sutun, order = df[sutun].value_counts().index)
  plt.title(f"{sutun} için count plot")
  plt.xlabel(sutun)
  plt.ylabel("Frekans")
  plt.xticks(rotation=45)
  plt.show()
````

# Pivot Tablo Analizi

### Amaç
Veri setini iki yönlü kırılımlar ile özetleyerek pivot tablo mantığını kavramak, karşılaştırmalı analizler yapmak ve ortaya çıkan sonuçları yorumlamak.

---

###Analiz Adımları

**1. Veri Setinin Yüklenmesi ve Hazırlanması**
Analize başlamadan önce veri seti ortama aktarılır. Veri setinde yer alan ve tekrar eden kayıtlar sebebiyle oluşan gereksiz isimsiz sütunlar (Unnamed) temizlenerek veri analize hazır hale getirilir.

**2. Kategori ve Müşteri Tipine Göre Pivot Tablo Analizi**
Ürün kategorileri ile müşteri profilleri (Bireysel, Kurumsal vb.) arasındaki ilişki incelenir. Hangi müşteri grubunun hangi kategorideki ürünlere daha fazla ilgi gösterdiği ve işlem hacmi oluşturduğu tespit edilir.

#### **3. Şehir ve Ödeme Türüne Göre Pivot Tablo Analizi**
Coğrafi konumlara göre müşterilerin tercih ettiği ödeme yöntemleri (Kredi Kartı, Havale, Kapıda Ödeme vb.) kırılıma uğratılır. Şehir bazında ödeme alışkanlıklarının değişip değişmediği analiz edilir.

#### **4. Kategori ve Ödeme Türüne Göre Pivot Tablo Analizi**
Satılan ürün grupları ile bu ürünler için kullanılan ödeme yöntemleri eşleştirilir. Belirli ürün kategorilerinde (örneğin yüksek fiyatlı ürün gruplarında) taksit veya farklı ödeme türlerinin ağırlığı incelenir.

#### **5. Şehir ve Müşteri Tipine Göre Pivot Tablo Analizi**
Farklı şehirlerdeki müşteri yapısı ortaya çıkarılır. Hangi illerde kurumsal müşteri yoğunluğu olduğu, hangi illerde bireysel satışların öne çıktığı belirlenerek bölgesel müşteri dağılımı özetlenir.

---

### **Sonuçların Karşılaştırılması ve Yorumlanması**
Oluşturulan iki yönlü tablolar üzerinden en yüksek ve en düşük yoğunluğa sahip kesişim noktaları belirlenir. Çıkan sonuçlar doğrultusunda müşteri eğilimleri ve operasyonel durumlara yönelik stratejik yorumlar yapılır.


## 1. Neden PIVOT TABLE Kullanıyoruz?

* Veriyi Özetleme: Binlerce satırlık ham veriyi büyük resme dönüştürmek için kullanıyoruz.
* Hızlı Karşılaştırma: Kategorik değişkenlerin (örn: bölge, yıl, segment) birbiriyle ilişkisini anında görmek için tercih ediyoruz.
* Agregasyon: Veriyi gruplayıp aynı anda toplam, ortalama veya sayı (count) gibi metrikleri hızlıca hesaplamamızı sağlıyor.

## 2. Mantıksal Yapısı (Nasıl Kuruyoruz?)
Pivot tabloyu oluştururken şu 4 temel parametreyi ekiyoruz:
```python
index: Satırda görmek istediğimiz kategorik değişken.
columns: Sütunda görmek istediğimiz kategorik değişken.
values: İçeride hesaplayacağımız sayısal sütun.
aggfunc: Hesaplama yöntemimiz ('mean', 'sum', 'count' vb.).
```
## 3. Kod Pratiğimiz
```python
import pandas as pd
# Örnek DataFrame (Hatırlatma amaçlı hızlı veri setimiz)

df = pd.DataFrame({
    'Yil':,
    'Bolge': ['Marmara', 'Ege', 'Marmara', 'Marmara', 'Ege', 'Ege'],
    'Satis':,
    'Gelir': [5000, 7000, 9000, 5500, 8500, 12000]
})
```
# SENARYO 1: Temel Pivot (Yıl ve Bölge bazında Toplam Satış)

```python
pivot_temel = df.pivot_table(
    index='Yil',
    columns='Bolge',
    values='Satis',
    aggfunc='sum'
)
print(pivot_temel)
```
# SENARYO 2: İleri Seviye Pivot (Çoklu Metrik ve Genel Toplamlar)
```python
pivot_gelismis = df.pivot_table(
    index='Yil',
    columns='Bolge',
    values=['Satis', 'Gelir'],
    aggfunc={'Satis': 'sum', 'Gelir': 'mean'}, # Satış için toplam, gelir için ortalama aldık
    fill_value=0,                               # Boş (NaN) kalan hücrelere 0 bastık
    margins=True                                # Sağa ve alta "All" (Genel Toplam) ekledik
)
print(pivot_gelismis)
```

````python
df = pd.read_csv("e_ticaret_veri_seti_aykiri_degerler_duzenlendi.csv")
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

df.head()
````

````python
# kategori ve müsteri tipine gore pivot tablo analizi
# her kategori icin musteri tiplerine goe ortalama toplam tutar

pivot_kategori_musteri = pd.pivot_table(
    df,
    index = "kategori",
    columns= "musteri_tipi",
    values = "toplam_tutar",
    aggfunc = "mean"
).round(1)

pivot_kategori_musteri
````

````python
# sehir ve odeme turu gore pivot tablo analizi
# her sehir icin odeme turune  gore  toplam tutar

pivot_sehir_odeme_turu = pd.pivot_table(
    df,
    index = "sehir",
    columns= "odeme_turu",
    values = "toplam_tutar",
    aggfunc = "sum"
).round(1)

pivot_sehir_odeme_turu
````

````python
# kategori ve odeme turune gore pivot tablo analizi
pivot_kategori_odeme = pd.pivot_table(
    df,
    index = "kategori",
    columns = "odeme_turu",
    values = "musteri_puani",
    aggfunc = "mean"
).round(2)

pivot_kategori_odeme
````

````python
# sehir ve musteri tipine gore pivot table analizi

pivot_sehir_musteri_tipi = pd.pivot_table(
    df,
    index = "sehir",
    columns = "musteri_tipi",
    values = "kargo_ucreti",
    aggfunc = "mean"
).round(1)

pivot_sehir_musteri_tipi
````
